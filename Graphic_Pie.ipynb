```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_excel("StudentsPrepared.xlsx")
```

```python
df.head()
```

```python
set(df['Target'])
```

```python
#Calcular o total de alunos(as) por tipo de status
df_targets_percent = df.groupby('Target')['Target'].count()
```

```python
print("\ndf_targets_percent", df_targets_percent)
```

```python
total_estudantes = len(df)
```

```python
print("\ntotal_estudantes", total_estudantes)
```

```python
df_target_porcentagem = df_targets_percent / total_estudantes * 100 #ou df_target_porcentagem = df_targets_percent / total_estudantes * 100
```

```python
print("\ndf_target_porcentagem", df_target_porcentagem)
```

```python
#Separando os valores e nomes em uma lista, para deixar o gráfico mais apresentável
labels = df_target_porcentagem.index.tolist()
```

```python
print("labels", labels )
```

```python
sizes = df_target_porcentagem.values.tolist()
```

```python
print("sizes", sizes)
```

```python
#Criando o gráfico
plt.style.use('dark_background')
figura, grafico = plt.subplots(figsize=(18, 6))
grafico.pie(sizes, autopct='%1.0f%%', colors=[ '#e34c42','#4dc471','#3b71db'], labeldistance = 1.1,
           explode=[0, 0, .4])
grafico.axis('equal')
plt.title('Porcentagem por status target')
plt.legend(labels, loc='best')
plt.show()
```

```python
# Cores opcionais para deixar mais visual
colors = ['#ff9999','#66b3ff','#99ff99']

# Cria o gráfico de pizza
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)

# Deixa o gráfico redondo
plt.axis('equal')
plt.title('Distribuição dos Estudantes por Situação')

# Mostra o gráfico
plt.show()
```

autopct='%1.1f%%' mostra as porcentagens no gráfico.

startangle=140 gira o início da pizza para melhor visualização.

plt.axis('equal') garante que a pizza fique redonda, não oval.
